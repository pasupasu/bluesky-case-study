## How To Interpret:
- `interactions.csv`: Contains the anonymized follower edge list. Each row consists of two comma-separated integers representing a directed following relation (i.e., user u follows user v).
- `followers.csv`: Contains the anonymized interactions edge list. Each row consists of six comma-separated integers representing a comment, repost, or quote interaction. These integers correspond to the following fields, in this order: user_id, replied_author, thread_root_author, reposted_author,quoted_author, and date.

```python
import os
from bluesky_case_study.paths import DATA_DIR
import polars as pl
```

```python
followers_df = pl.scan_csv(
    DATA_DIR / "followers.csv",
    has_header= False,
    new_columns = ["follower_id", "followed_id"]
).collect()

followers_df.shape
followers_df.head(); followers_df.sample(10)   # random rows are more honest than the first rows
```

```python
interactions_df = pl.scan_csv(
    DATA_DIR / "interactions.csv",
    has_header = False,
    new_columns = ["user_id", "replied_author", "thread_root_author", "reposted_author", "quoted_author", "date"]
).collect()

interactions_df.shape
interactions_df.head(); interactions_df.sample(10)
```
